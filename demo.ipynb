# AI Fitness Trainer

This notebook demonstrates how to use the specific exercise detection modules.

## Instructions
1. Run the cell below to import dependencies.
2. Choose an exercise by setting the `EXERCISE` variable.
3. Run the main loop cell to start the webcam feed.

```python
import cv2
import mediapipe as mp
import numpy as np

# Import exercise modules
from pushups import PushUpDetector
from squats import SquatDetector
from crunches import AbCrunchDetector
from plank import PlankTimer
from lunges import LungeDetector
from bicep_curls import BicepCurlDetector
from jumping_jacks import JumpingJackDetector
from high_knees import HighKneeDetector
```

```python
# Select Exercise: 'pushup', 'squat', 'crunch', 'plank', 'lunge', 'bicep_curl', 'jumping_jack', 'high_knee'
EXERCISE = 'pushup' 

if EXERCISE == 'pushup':
    detector = PushUpDetector()
elif EXERCISE == 'squat':
    detector = SquatDetector()
elif EXERCISE == 'crunch':
    detector = AbCrunchDetector()
elif EXERCISE == 'plank':
    detector = PlankTimer()
elif EXERCISE == 'lunge':
    detector = LungeDetector()
elif EXERCISE == 'bicep_curl':
    detector = BicepCurlDetector()
elif EXERCISE == 'jumping_jack':
    detector = JumpingJackDetector()
elif EXERCISE == 'high_knee':
    detector = HighKneeDetector()
else:
    print("Invalid exercise selected")
```

```python
mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose

cap = cv2.VideoCapture(0)

# Setup mediapipe instance
with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:
    while cap.isOpened():
        ret, frame = cap.read()
        
        if not ret:
            break
        
        # Recolor image to RGB
        image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image.flags.writeable = False
      
        # Make detection
        results = pose.process(image)
    
        # Recolor back to BGR
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
        
        # Extract landmarks
        try:
            landmarks = results.pose_landmarks.landmark
            
            # Process using the selected detector
            image = detector.process(image, landmarks)
            
        except Exception as e:
            # print(e) # Pass if no landmarks detected
            pass
        
        
        # Render detections
        mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS,
                                mp_drawing.DrawingSpec(color=(245,117,66), thickness=2, circle_radius=2), 
                                mp_drawing.DrawingSpec(color=(245,66,230), thickness=2, circle_radius=2) 
                                 )               
        
        cv2.imshow('AI Fitness Trainer', image)

        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```
